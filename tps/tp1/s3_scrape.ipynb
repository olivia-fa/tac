# Scraping: récupération des PDF

## Imports

```python
import os
import re
import time
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

## Récupération de tous les PDF



```python
headers = {'User-Agent': 'Mozilla/5.0 (Macintosh; Intel Mac OS X 10_10_1) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/39.0.2171.95 Safari/537.36'}

# Créer liste
pdf_links = []

# Chercher le contenu de la page
root_url = f"https://max.de.wilde.web.ulb.be/camille/"
response = requests.get(root_url, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')

# Parcourir les liens
for link in soup.find_all("a"):
    href = link.get("href")
    if href and href.endswith(".pdf") :
        title = link.text.strip()
        url = root_url + link.get('href')
        pdf_links.append([url, title])
```

```python
# Affichage du nombre de pdf trouvés
len(pdf_links)
```

```python
# Affichage des 10 premières entrées
pdf_links[:10]
```

## Téléchargement des pdfs


```python
# Prendre le contenu
for url, title in pdf_links:
    pdf_contenu = requests.get(url, headers=headers)
    file_path = "pdfs/" + title
    
    # Enregistrer les pdfs
    with open(file_path, "wb") as writer:
        writer.write(pdf_contenu.content)
```
