## Imports

```python
import json
import requests
```

## Utiliser [Europeana](https://www.europeana.eu/fr) pour obtenir des informations sur le patrimoine culturel européen
Europeana est la plateforme numérique européenne dédiée au patrimoine culturel.

Api reference : https://www.europeana.eu/fr


Il faut s'enregistrer pour créer une clé API 

### Recherche de documents sur Bruxelles (1890-1920)

```python

query = "Brussels"

# Recherche de documents historiques 
headers = {'X-Api-Key': 'baconset'}
url = "https://api.europeana.eu/record/v2/search.json"
data = {
    'query': 'Brussels',
    # Tranche temporelle specifique
    'qf': 'YEAR:[1890 TO 1920]',
    # Nombre de resultats
    'rows': 10
}

resp = requests.get(url, data, headers=headers)

try:
    json_list = json.loads(resp.text)
except json.JSONDecodeError:
    print(resp.text)

for item in json_list['items']:
    title = item.get('title', ['Titre inconnu'])
    year = item.get('year', ['Date inconnue'])
    link = item.get('guid', 'Lien inconnu')

    print(f"Titre: {title[0]}")
    print(f"Annee: {year[0] if year else 'Date inconnue'}")
    print(f"Lien : {link}")

```
